##Match/case
La estructura match evalua una variable frente a varios bloques case. El comodin case __ funciona como ninguna de las anteriores.

sintaxis
```
match (condicion):
  case v1: aqui se ejecuta si condicion==v1
  case v2: aqui se ejecuta si condicion==v2
  case v3: aqui se ejecuta si condicion==v3
  case __: ninguna de las anteriores
```


````python
helado = input("quieres helado?")
````

````python
#if/else
if helado.lower()=="si" or helado.lower()=="sí":
  print("te lo compro")
else: print("no te lo compro")
````

````python
#if/elif/else
if helado.lower()=="si":
  print("te lo compro")
elif helado.lower()=="sí":
  print("te lo compro")
elif helado.lower()=="sípi":
  print("te lo compro")
else: print("no te lo compro")
````

````python
#match/case
match helado.lower():
  case "si": print("no te lo compro")
  case "sí": print("no te lo compro")
  case "sipis": print("no te lo compro")
  case "clarinetes": print("no te lo compro")
  case __: print("no te lo compro")
````
