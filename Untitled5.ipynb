````python
flores = {
    "tipo": ["rosa", "tulipan", "girasol", "amapola", "lirio"],
    "cantidad": [1, 17, 3, 8, 5],
    "precio": [4000, 80000, 5000, 15000, 23000]
}
````

````python
flores = ["rosa", "tulipan", "girasol", "amapola", "lirio"]
cantidad = [1, 17, 3, 8, 5]
precio = [4000, 80000, 5000, 15000, 23000]

carrito_flores = []
carrito_cantidades = []
carrito_precios = []
carrito_subtotales = []

print("Bienvenido a mi floristeria\n")

opc = 0
while opc != 5:
    print("\nMenú:")
    print("1. Consultar flores")
    print("2. Comprar flores")
    print("3. Pagar")
    print("4. Factura de compra")
    print("5. Salir")

    opc = int(input("Ingrese su opción: "))

    if opc == 1:
        print("\nEstas son nuestras flores disponibles:")
        print("N° | Flores | Cantidad | Precio")
        for i in range(len(flores)):
            print(i + 1, "|", flores[i], "|", cantidad[i], "| $", precio[i])

    elif opc == 2:
        print("\n--- Comprar Flores ---")
        for i in range(len(flores)):
            print(i + 1, ".", flores[i], "(Disponibles:", cantidad[i], "Precio: $", precio[i], ")")

        seleccion = int(input("Seleccione el número de la flor que desea comprar: ")) - 1

        if seleccion >= 0 and seleccion < len(flores):
            cant_deseada = int(input("¿Cuántas unidades desea? "))

            if cant_deseada <= 0:
                print("La cantidad debe ser mayor a 0.")
            elif cant_deseada <= cantidad[seleccion]:
                cantidad[seleccion] = cantidad[seleccion] - cant_deseada
                subtotal = cant_deseada * precio[seleccion]

                carrito_flores.append(flores[seleccion])
                carrito_cantidades.append(cant_deseada)
                carrito_precios.append(precio[seleccion])
                carrito_subtotales.append(subtotal)

                print("¡Se han agregado al carrito!")
            else:
                print("No hay suficiente stock disponible.")
        else:
            print("Selección inválida.")

    elif opc == 3:
        if len(carrito_flores) == 0:
            print("\nEl carrito está vacío. Agregue flores primero.")
        else:
            total = 0
            for sub in carrito_subtotales:
                total = total + sub

            print("\nProcesando el pago de: $", total)
            print("¡Pago realizado con éxito!")

            carrito_flores.clear()
            carrito_cantidades.clear()
            carrito_precios.clear()
            carrito_subtotales.clear()

    elif opc == 4:
        print("\nFactura de Compra")
        if len(carrito_flores) == 0:
            print("No tienes productos en el carrito actualmente.")
        else:
            total = 0
            print("Flor | Cant. | P. Unitario | Subtotal")
            print("----------------------------------------")
            for i in range(len(carrito_flores)):
                print(carrito_flores[i], "|", carrito_cantidades[i], "| $", carrito_precios[i], "| $", carrito_subtotales[i])
                total = total + carrito_subtotales[i]
            print("----------------------------------------")
            print("TOTAL A PAGAR: $", total)

    elif opc == 5:
        print("Gracias por su visita. ¡Vuelva pronto!")

    else:
        print("Opción no válida")
````
