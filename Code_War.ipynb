```python
agenda = []
nombre = input("INGRESE EL NOMBRE QUE LE GUSTARIA TENER EN SU AGENDA: ")
while True:
  print(f"BIENVENIDO A {nombre}")
  print("1. Ver agenda")
  print("2. Agregar actividad")
  print("3. Buscar actividad")
  print("4. Eliminar actividad")
  print("5. Mostrar cantidad de actividades")
  print("6. Ver primera actividad")
  print("7. Ver última actividad")
  print("8. Ordenar agenda (A-Z)")
  print("9. Salir")

  opcion = input("Selecciona una opción (1-9): ")
  if opcion == "1":
      if not agenda:
          print("La agenda está vacía.")
      else:
          print("Listado de actividades:")
          for posicion, actividad in enumerate(agenda, 1):
              print(f"{posicion}. {actividad}")
  elif opcion == "2":
      nueva = input("Escribe la nueva actividad: ")
      agenda.append(nueva)
      print(f"'{nueva}' se agregó a la agenda!")
  elif opcion == "3":
      busqueda = input("¿Qué actividad buscas?: ")
      if busqueda in agenda:
        print(f"Sí '{busqueda}' está en tu agenda.")
      else:
        print(f"No se encontró '{busqueda}' en la agenda.")
  elif opcion == "4":
    eliminar = input("Escribe la actividad que quieres borrar: ")
    if eliminar in agenda:
      agenda.remove(eliminar)
      print(f"Se eliminó '{eliminar}' correctamente.")
    else:
      print(f"No se pudo eliminar. '{eliminar}' no existe.")
  elif opcion == "5":
    print(f"Tienes {len(agenda)} actividad(es) en total.")
  elif opcion == "6":
      if not agenda:
        print("La agenda está vacía.")
      else:
        print(f"La primera actividad es: {agenda[0]}")
  elif opcion == "7":
    if not agenda:
      print("La agenda está vacía.")
    else:
      print(f"La última actividad es: {agenda[-1]}")
  elif opcion == "8":
    if not agenda:
       print("La agenda está vacía.")
    else:
      agenda.sort()
      print("Agenda ordenada alfabéticamente de la A a la Z.")
  elif opcion == "9":
      print("Hasta luego")
      break
  else:
      print("Opción inválida. Intenta del 1 al 9.")
```
